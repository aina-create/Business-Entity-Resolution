```python
import pandas as pd
import re

base_path = (
    "/Users/aiwinthomas/Downloads/"
    "Business Entity Resolution/dataset/train/"
)

def normalize_name(name):
    if pd.isna(name):
        return ""

    name = str(name).lower()
    name = re.sub(r"[^\w\s]", " ", name)
    name = re.sub(r"\s+", " ", name).strip()

    return name

# Load a small sample from each source
s1 = pd.read_csv(
    base_path + "train_source1.tsv",
    sep="\t",
    dtype=str,
    nrows=5000
)

s2 = pd.read_csv(
    base_path + "train_source2.tsv",
    sep="\t",
    dtype=str,
    nrows=5000
)

s3 = pd.read_csv(
    base_path + "train_source3.tsv",
    sep="\t",
    dtype=str,
    nrows=5000
)

# Normalize business names
for df in [s1, s2, s3]:
    df["name_normalized"] = df["business_name"].apply(
        normalize_name
    )

    df["name_prefix"] = df["name_normalized"].str[:4]

print("Source 1 sample:", len(s1))
print("Source 2 sample:", len(s2))
print("Source 3 sample:", len(s3))
```

```python
# Keep records with usable blocking keys
s1_valid = s1[
    s1["country"].notna()
    & s1["name_prefix"].ne("")
]

s2_valid = s2[
    s2["country"].notna()
    & s2["name_prefix"].ne("")
]

s3_valid = s3[
    s3["country"].notna()
    & s3["name_prefix"].ne("")
]

# Generate Source 1 -> Source 2 candidates
candidates_s2 = s1_valid.merge(
    s2_valid,
    on=["country", "name_prefix"],
    suffixes=("_s1", "_s2")
)

# Generate Source 1 -> Source 3 candidates
candidates_s3 = s1_valid.merge(
    s3_valid,
    on=["country", "name_prefix"],
    suffixes=("_s1", "_s3")
)

print("Source 1 -> Source 2 candidates:", len(candidates_s2))
print("Source 1 -> Source 3 candidates:", len(candidates_s3))
print(
    "Total candidates:",
    len(candidates_s2) + len(candidates_s3)
)
```

```python
# Load ground truth for the Source 1 sample
gt = pd.read_csv(
    base_path + "train_ground_truth.tsv",
    sep="\t",
    dtype=str
)

sample_s1_ids = set(s1["entity_id"])

gt_sample = gt[
    gt["source1_entity_id"].isin(sample_s1_ids)
].copy()

# Convert candidate pairs into sets of ID pairs
candidate_pairs_s2 = set(
    zip(
        candidates_s2["entity_id_s1"],
        candidates_s2["entity_id_s2"]
    )
)

candidate_pairs_s3 = set(
    zip(
        candidates_s3["entity_id_s1"],
        candidates_s3["entity_id_s3"]
    )
)

candidate_pairs = candidate_pairs_s2 | candidate_pairs_s3

# Count true matches found among the candidates
true_matches = 0
found_matches = 0

for _, row in gt_sample.iterrows():
    s1_id = row["source1_entity_id"]

    matched_ids = [
        x.strip()
        for x in row["matched_entity_ids"].split(",")
        if x.strip()
    ]

    for matched_id in matched_ids:
        true_matches += 1

        if (s1_id, matched_id) in candidate_pairs:
            found_matches += 1

print("Ground-truth matches for sampled Source 1 records:", true_matches)
print("Matches found by blocking:", found_matches)

if true_matches:
    print(
        "Candidate recall:",
        round(found_matches / true_matches * 100, 2),
        "%"
    )
```

```python
def find_records(file_path, needed_ids, chunksize=100_000):
    found = []

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        dtype=str,
        chunksize=chunksize
    ):
        matches = chunk[chunk["entity_id"].isin(needed_ids)]

        if not matches.empty:
            found.append(matches)

            # Stop if every requested ID has been found
            found_ids = set(
                pd.concat(found)["entity_id"]
            )

            if found_ids >= needed_ids:
                break

    if found:
        return pd.concat(found, ignore_index=True)

    return pd.DataFrame(
        columns=[
            "entity_id",
            "business_name",
            "business_address",
            "country"
        ]
    )


s2_true_records = find_records(
    base_path + "train_source2.tsv",
    needed_s2_ids
)

s3_true_records = find_records(
    base_path + "train_source3.tsv",
    needed_s3_ids
)

print("Source 2 records found:", len(s2_true_records))
print("Source 3 records found:", len(s3_true_records))
```

```python
import re
import unicodedata
import pandas as pd

LEGAL_WORDS = {
    "private", "limited", "ltd", "pvt",
    "llc", "llp", "inc", "incorporated",
    "corporation", "corp", "company", "co",
    "pllc", "m/s"
}

def normalize_for_blocking(value):
    if pd.isna(value):
        return ""

    value = str(value).lower()

    # Normalize accents: Ówl -> owl
    value = unicodedata.normalize("NFKD", value)
    value = "".join(
        ch for ch in value
        if not unicodedata.combining(ch)
    )

    # Replace punctuation with spaces
    value = re.sub(r"[^\w\s]", " ", value)
    value = re.sub(r"\s+", " ", value).strip()

    return value

def get_name_tokens(value):
    normalized = normalize_for_blocking(value)

    return {
        token
        for token in normalized.split()
        if len(token) >= 3 and token not in LEGAL_WORDS
    }

# Inspect tokenization on our known examples
for name in [
    "Novent Owl PLLC",
    "PLLC Novent Ówl",
    "Haven Exim",
    "M/s Haven Exim",
    "Anderson and Baker Adamas",
]:
    print(name, "->", get_name_tokens(name))
```

```python
def retrieve_token_candidates(file_path, test_tokens, chunksize=100_000):
    candidate_records = []

    for chunk in pd.read_csv(
        file_path,
        sep="\t",
        dtype=str,
        chunksize=chunksize
    ):
        chunk["tokens"] = chunk["business_name"].apply(
            get_name_tokens
        )

        mask = chunk["tokens"].apply(
            lambda tokens: bool(tokens & test_tokens)
        )

        if mask.any():
            candidate_records.append(chunk.loc[mask])

    if candidate_records:
        return pd.concat(candidate_records, ignore_index=True)

    return pd.DataFrame()


s2_token_candidates = retrieve_token_candidates(
    base_path + "train_source2.tsv",
    test_tokens
)

print("Source 2 token candidates:", len(s2_token_candidates))
```

```python
from collections import Counter

def token_frequency(series):
    counts = Counter()

    for value in series.dropna():
        counts.update(get_name_tokens(value))

    return counts

s1_address_frequency = token_frequency(s1["business_address"])
s2_address_frequency = token_frequency(s2["business_address"])
s3_address_frequency = token_frequency(s3["business_address"])

print("Most common Source 2 address tokens:")
print(s2_address_frequency.most_common(20))

print("\nMost common Source 3 address tokens:")
print(s3_address_frequency.most_common(20))
```

```python
ADDRESS_STOP_WORDS = {
    "road", "rd", "street", "st", "floor",
    "drive", "dr", "avenue", "ave", "lane",
    "ln", "city", "unit", "suite", "near",
    "plot", "block", "building", "floor",
    "new", "north", "south", "east", "west",
    "and", "the", "for", "of", "in",
    "null", "none", "nan"
}

def get_address_tokens(value):
    tokens = get_name_tokens(value)

    return {
        token for token in tokens
        if token not in ADDRESS_STOP_WORDS
        and not token.isdigit()
    }
```
